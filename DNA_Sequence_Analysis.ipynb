# DNA Sequence Analysis

## Project Objective

This project analyzes a DNA sequence using Python. The analysis includes:

- DNA sequence length
- Nucleotide composition (A, T, G, C)
- GC content
- Start codons
- Stop codons

## Step 1: Store the DNA Sequence

In this step, we store a DNA sequence in a Python variable for analysis.

```python
# Store DNA Sequence 
dna = "ATGCGATACGATAGCTAGCTAA"
print("DNA Sequence:")
print(dna)
```

## Step 2: Calculate DNA Length 
The length of a DNA Sequence tells us the total number of nucleotides present.

```python
# Calculate DNA length

length = len(dna)

print("Length of DNA Sequence:", length)
```

## Step 3: Count Nucleotides

In this step, we count the number of each nucleotide (A, T, G, and C) present in the DNA sequence. This helps us understand the nucleotide composition of the DNA.

```python
# Count nucleotides
A = dna.count("A")
T = dna.count("T")
G = dna.count("G")
C = dna.count("C")
print("A:", A)
print("T:", T)
print("G:", G)
print("C:", C)
```

## Step 4: Calculate GC Content

GC content is the percentage of guanine (G) and cytosine (C) nucleotides in a DNA sequence. It is an important characteristic of DNA and is often used in genetics, genomics, and bioinformatics to study DNA stability and compare different organisms.

```python
# Calculate GC Content 
gc_content = ( (G + C)/ length) * 100
print(f"GC Content: {gc_content:.2f}%")
```

## Step 5: Identify the Start Codon
A start codon marks the beginning of protein syntheis in a DNA Sequence. The most common start codon is **ATG**, which codes for the amino acid methionine. In this step, we check whether the DNA sequence begins with the tart codon.

```python
# Check for Start Codon

if dna.startswith("ATG"):
    print("Start codon (ATG) found at the beginning of the DNA sequence.")
else:
    print("Start codon (ATG) not found at the beginning of the DNA sequence.")
```

## Step 6: Identify the Stop Codon

A stop codon signals the end of protein synthesis. The three common stop codons in DNA are **TAA**, **TAG**, and **TGA**. In this step, we check whether the DNA sequence ends with one of these stop codons.

```python
# Check for Stop Codon

if dna.endswith(("TAA", "TAG", "TGA")):
    print("Stop codon found at the end of the DNA sequence.")
else:
    print("No stop codon found at the end of the DNA sequence.")
```

## Step 7: Reverse the DNA Sequence

In this step, we reverse the DNA sequence. Reversing a DNA sequence is a common operation in bioinformatics and is useful for sequence analysis and understanding DNA structure.

```python
# Reverse DNA Sequence

reverse_dna = dna[::-1]

print("Original DNA Sequence:", dna)
print("Reversed DNA Sequence:", reverse_dna)
```

## Step 8: Generate the Complementary DNA Sequence 
DNA is made up of complementary base pairs.
- Adenine (A) pairs with Thymine (T)
- Thymine (T) pairs with Adenine (A)
- Guanine (G) pairs with Cytosine (C)
- Cytosine (C) pairs with Guranine (G)

  In this step, we generate the complementary DNA sequence using Python.

```python
# Generate Complementary DNA Sequence 

complement = " "

for base in dna: 
    if base == "A":
        complement += "T"
    elif base == "T":
        complement += "A"
    elif base == "G":
        complement += "C"
    elif base == "C":
        complement += "G"

print("Original DNA Sequence :", dna)
print("Complement DNA Sequence:", complement)
```

## Step 9: Generate the Reverse Complement DNA Sequence

The reverse complement of a DNA sequence is obtained by first generating the complementary DNA sequence and then reversing it. This sequence is widely used in bioinformatics for sequence analysis, genome annotation, and primer design.

```python
# Generate Reverse Complement DNA Sequence

reverse_complement = complement[::-1]

print("Original DNA Sequence       :", dna)
print("Complement DNA Sequence     :", complement)
print("Reverse Complement Sequence :", reverse_complement)
```

# Step 10: DNA Sequence Analysis Report

This section summarizes all the analyses performed on the DNA sequence. It provides a concise overview of the sequence characteristics, including its length, nucleotide composition, GC content, and the presence of start and stop codons.

```python
# DNA Sequence Analysis Report

print("=" * 50)
print("        DNA SEQUENCE ANALYSIS REPORT")
print("=" * 50)

print("\nOriginal DNA Sequence:")
print(dna)

print("\nSequence Length:", length)

print("\nNucleotide Composition")
print("----------------------")
print("Adenine (A):", A)
print("Thymine (T):", T)
print("Guanine (G):", G)
print("Cytosine (C):", C)

print(f"\nGC Content: {gc_content:.2f}%")

if dna.startswith("ATG"):
    print("\nStart Codon: Present (ATG)")
else:
    print("\nStart Codon: Not Present")

if dna.endswith(("TAA", "TAG", "TGA")):
    print("Stop Codon: Present")
else:
    print("Stop Codon: Not Present")

print("\nReverse DNA Sequence:")
print(reverse_dna)

print("\nComplement DNA Sequence:")
print(complement)

print("\nReverse Complement DNA Sequence:")
print(reverse_complement)

print("\nProject Status: Analysis Completed Successfully!")

print("=" * 50)
```
